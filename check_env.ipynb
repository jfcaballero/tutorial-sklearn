```python
from __future__ import print_function
from packaging.version import Version
import sys

# Detectar colores en terminal (opcional)
try:
    import curses
    curses.setupterm()
    assert curses.tigetnum("colors") > 2
    OK = "\x1b[1;%dm[ OK ]\x1b[0m" % (30 + curses.COLOR_GREEN)
    FAIL = "\x1b[1;%dm[FAIL]\x1b[0m" % (30 + curses.COLOR_RED)
except:
    OK = '[ OK ]'
    FAIL = '[FAIL]'

import importlib

# Función para comprobar versiones de librerías
def import_version(pkg, min_ver, fail_msg=""):
    try:
        mod = importlib.import_module(pkg)
        ver = mod.__version__
        if Version(ver) < Version(min_ver):
            print(FAIL, "%s version %s or higher required, but %s installed."
                  % (pkg, min_ver, ver))
        else:
            print(OK, '%s version %s' % (pkg, ver))
    except ImportError:
        print(FAIL, '%s not installed. %s' % (pkg, fail_msg))

# Mostrar versión de Python
print('Using python in', sys.prefix)
print(sys.version)

# Limpiar versión de Python (solo el número)
clean_version = sys.version.split()[0]  # '3.12.12'
pyversion = Version(clean_version)

# Comprobar versión mínima
if pyversion < Version("3.8"):
    print(FAIL, "Python version 3.8 is required,"
                " but %s is installed." % clean_version)

print()

# Requisitos de librerías
requirements = {
    'numpy': "1.20", 'scipy': "1.6", 'matplotlib': "3.3",
    'IPython': "7.22", 'sklearn': "0.24", 'pandas': "1.2",
    'PIL': "8.2", 'ipympl': '0.7'
}

# Comprobar cada librería
for lib, required_version in requirements.items():
    import_version(lib, required_version)
```
